## 4. Find a String
[https://www.hackerrank.com/challenges/find-a-string/problem](https://www.hackerrank.com/challenges/find-a-string/problem)

```python
# Sample input (simulated stdin)
_data = iter('ABCDCDC\nCDC'.split("\n"))
input = lambda: next(_data)

def count_substring(string, sub_string):
    count = 0
    for i in range(len(string) - len(sub_string) + 1):
        if string[i:i+len(sub_string)] == sub_string:
            count += 1
    return count

string = input().strip()
sub_string = input().strip()
print(count_substring(string, sub_string))
```

## 16. Collections.Counter()
[https://www.hackerrank.com/challenges/collections-counter/problem](https://www.hackerrank.com/challenges/collections-counter/problem)

```python
# Sample input (simulated stdin)
_data = iter('10\n2 3 4 5 6 8 7 6 5 18\n6\n6 55\n6 45\n6 55\n4 40\n18 60\n10 50'.split("\n"))
input = lambda: next(_data)

from collections import Counter

x = int(input())
shoes = Counter(map(int, input().split()))
earned = 0
for _ in range(int(input())):
    size, price = map(int, input().split())
    if shoes[size] > 0:
        earned += price
        shoes[size] -= 1
print(earned)
```

## 18. The Minion Game
[https://www.hackerrank.com/challenges/the-minion-game/problem](https://www.hackerrank.com/challenges/the-minion-game/problem)

```python
# Sample input (simulated stdin)
_data = iter('BANANA'.split("\n"))
input = lambda: next(_data)

def minion_game(string):
    vowels = "AEIOU"
    n = len(string)
    stuart = kevin = 0
    for i, ch in enumerate(string):
        if ch in vowels:
            kevin += n - i
        else:
            stuart += n - i
    if stuart > kevin:
        print("Stuart", stuart)
    elif kevin > stuart:
        print("Kevin", kevin)
    else:
        print("Draw")

s = input()
minion_game(s)
```

## 19. Merge the Tools!
[https://www.hackerrank.com/challenges/merge-the-tools/problem](https://www.hackerrank.com/challenges/merge-the-tools/problem)

```python
# Sample input (simulated stdin)
_data = iter('AABCAAADA\n3'.split("\n"))
input = lambda: next(_data)

def merge_the_tools(string, k):
    for i in range(0, len(string), k):
        seen = ""
        for ch in string[i:i+k]:
            if ch not in seen:
                seen += ch
        print(seen)

string, k = input(), int(input())
merge_the_tools(string, k)
```

## 20. Piling Up!
[https://www.hackerrank.com/challenges/piling-up/problem](https://www.hackerrank.com/challenges/piling-up/problem)

```python
# Sample input (simulated stdin)
_data = iter('2\n6\n4 3 2 1 3 4\n3\n1 3 2'.split("\n"))
input = lambda: next(_data)

from collections import deque

for _ in range(int(input())):
    n = int(input())
    d = deque(map(int, input().split()))
    prev = float("inf")
    ok = True
    while d:
        cur = d.popleft() if d[0] >= d[-1] else d.pop()
        if cur > prev:
            ok = False
            break
        prev = cur
    print("Yes" if ok else "No")
```
