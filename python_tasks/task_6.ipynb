## 12. Set .discard(), .remove() & .pop()

**Task:** Read a set and execute N commands: pop, remove x, or discard x. Print the final sum.

**Sample output guide:** Use the HackerRank sample input to verify the final sum. Note: set.pop() removes an arbitrary element.

**HackerRank:** Search the exact problem title on HackerRank or use the link from your question.

```python
n = int(input())
s = set(map(int, input().split()))
for _ in range(int(input())):
    command = input().split()
    if command[0] == "pop":
        if s:
            s.pop()
    elif command[0] == "remove":
        s.remove(int(command[1]))
    elif command[0] == "discard":
        s.discard(int(command[1]))
print(sum(s))
```

## 17. Check Strict Superset

**Task:** Print True if set A is a strict superset of every other given set; otherwise print False.

## 18. Set Mutations

**Task:** Read set A, then apply each operation with another set and print the sum of A.

**Sample output guide:** Use the HackerRank sample input to verify the final sum.

**HackerRank:** Search the exact problem title on HackerRank or use the link from your question.

```python
n = int(input())
a = set(map(int, input().split()))
for _ in range(int(input())):
    operation, _length = input().split()
    other = set(map(int, input().split()))
    if operation == "intersection_update":
        a.intersection_update(other)
    elif operation == "update":
        a.update(other)
    elif operation == "symmetric_difference_update":
        a.symmetric_difference_update(other)
    elif operation == "difference_update":
        a.difference_update(other)
print(sum(a))
```

## 20. Validating the Phone Number

**Task:** Read T strings and print Valid if each is exactly 10 digits and starts with 7, 8, or 9; otherwise print Invalid.

**Sample output guide:** Input: `2` then `9876543210` then `1234567890`  → Output: `Valid` then `Invalid`

**HackerRank:** Search the exact problem title on HackerRank or use the link from your question.

```python
import re

t = int(input())
for _ in range(t):
    phone = input().strip()
    print("Valid" if re.fullmatch(r"[789]\d{9}", phone) else "Invalid")
```
