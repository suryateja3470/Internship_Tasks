```python
N = int(input())
items = []
```

````python
items = []
commands = [
    ("insert", 0, 5), ("insert", 1, 10), ("insert", 0, 6),
    ("print",), ("remove", 6), ("append", 9), ("append", 1),
    ("sort",), ("print",), ("pop",), ("reverse",), ("print",)
]
for command in commands:
    op = command[0]
    if op == "insert":
        items.insert(command[1], command[2])
    elif op == "print":
        print(items)
    elif op == "remove":
        items.remove(command[1])
    elif op == "append":
        items.append(command[1])
    elif op == "sort":
        items.sort()
    elif op == "pop":
        items.pop()
    elif op == "reverse":
        items.reverse()
````

## 19. Collections.Counter()

```python
from collections import Counter

x = int(input())
sizes = Counter(map(int, input().split()))
customers = int(input())
earnings = 0

```


### Run a sample in Colab
The next cell uses sample values directly, so it runs without asking for input.

````python
from collections import Counter
sizes = Counter([2, 3, 4, 5, 6, 8, 7, 6, 5, 18])
orders = [(6, 55), (6, 45), (6, 55), (4, 40), (18, 60), (10, 50)]
earnings = 0
for size, price in orders:
    if sizes[size] > 0:
        earnings += price
        sizes[size] -= 1
print(earnings)
````
